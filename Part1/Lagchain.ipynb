```python
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import PromptTemplate
from langchain_core.prompts import ChatPromptTemplate
```

```python


llm = HuggingFacePipeline.from_model_id(
    model_id="Qwen/Qwen2.5-1.5B-Instruct",
    task="text-generation",
    pipeline_kwargs={
        "max_new_tokens": 2048},
)

# also use crumb/nano-mistral
```

```python
prompt = "What is the merge sort algorithm ? please write the code as well "
response = llm.invoke(prompt)
print(response)
# max_new_tokens will be are the responsible for generating the particular amount of tokens to be generated
```

```python
algo_tutor_template = """
You are an expert Computer science engineer. Explain the following algorithm: {algorithm_name}

Follow the EXACT structure:

### 1. The concept (what it is?)
Explain what problem this algorithm solves in 2 sentences.

### 2. The efficiency (Complexity analysis)
Define the time complexity and space complexity of the {algorithm_name}

### 3. Sample trace run
Show a simple step by step dry-run using the sample input: {sample_input}


Answer::
"""
```

```python
prompt = PromptTemplate(
    template=algo_tutor_template,
    input_variables=["algorithm_name", "sample_input"],
)
```

```python
final_prompt = prompt.format(algorithm_name = "Bubble Sort" , sample_input = [5,1,4,2])
print(final_prompt)
```

```python
chain = prompt | llm
# pipe operator -> take the prompt and feed it to the llm
# prompt is feeded inside the llm , creating the chain


result = chain.invoke({
    "algorithm_name" : "Binanry Search ",
    "sample_input" : [1,4,32,45]

})
print(result)
```

```python
template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a calculator that responds with reasoning about the calculation"),
        ("human", "Answer what is 10 multiplied by 20"),
        ("ai", "10 x 20 = 200, as it's equivalent to adding 10 to itself 20 times"),
        ("human", "{question}"),
    ]
)
```

```python
chain = template | llm
response = chain.invoke({"question" : "what is the square root of 100"})
print(response)
```
