```python
from langchain.agents import create_agent
from langchain_core.tools import tool
from langchain_groq import ChatGroq
from dotenv import load_dotenv
import os

load_dotenv()
```

```python
@tool
def get_weather(location :str) -> str:
    """Get the weather in a given location"""
    return f"The weather in {location} is sunny and 25 deg C"
```

```python
model = ChatGroq(
    model="openai/gpt-oss-120b",
    api_key=os.getenv("GROQ_API_KEY")
)
```

```python
agent = create_agent(
    model = model,
    tools = [get_weather],
    system_prompt = "You are a weather assistant that can answer questions and help with weather related tasks."
)
```

```python
result = agent.invoke({
    "messages": [
        {"role": "user", "content": "What is the weather in Tokyo?"}
    ]
})

print(result)
```

```python
for message in result["messages"]:
    print(message.pretty_print())
```

```python
result = agent.invoke({
    "messages": [
        {"role": "user", "content": "What is the weather in Tokyo compared to New York?"}
    ]
})

print(result)
```

```python
for messages in result["messages"]:
    print(messages.pretty_print())
```

```python
# actually it is making 2 tools here but to see completely use this 
for message in result["messages"]:
    print("=" * 60)
    print(type(message).__name__)
    print("Content:", message.content)

    if hasattr(message, "tool_calls") and message.tool_calls:
        print("Tool Calls:", message.tool_calls)
```
